```python
# When to use a 'with _ as _b:' expression

path='../text.txt'

with open(path, 'r') as f:
    text = f.read()

# is equivalent to:

f = open(path, 'r')

try:
    text = f.read()
finally:
    f.close()

# 'with' statement guarantees that the cleanup happens (even if an exception occurs)
# x ~ the context manager, and y is what object it returns (the alias is really a function call)

# under the hood 'with' uses 2 dunder methods:
# __enter()__
# __exit()__

# you can even define your own context manager 
class CustomContext():
    def __enter__(self):
        return self

    def __exit__(self, exc_type, exc, tb):
        print('All good')

with CustomContext() as x:
    print('inside context')
```

```python
# Linear classifier in pure python

import numpy as np

samples_per_class = 1500

negative_samples = np.random.multivariate_normal(
    mean=[0,3],
    cov = [[1, 0.5], [0.5, 1]], size = samples_per_class
)


positive_samples = np.random.multivariate_normal(
    mean=[3,0],
    cov = [[1, 0.5], [0.5, 1]],
    size = samples_per_class
)

print(positive_samples.shape)
print(positive_samples[:10])

plt.scatter(positive_samples[:, 0], positive_samples[:, 1])
plt.scatter(negative_samples[:, 0], negative_samples[:, 1])
plt.show()
```

```python
inputs = np.vstack((negative_samples, positive_samples)).astype(np.float32)

targets = np.vstack(
    (np.zeros((samples_per_class, 1), dtype="float32"),
    np.ones((samples_per_class, 1), dtype="float32"))
)

plt.scatter(inputs[:, 0], inputs[:, 1], c=targets[:, 0])

```

```python
input_dim = 2
output_dim = 1

W = tf.Variable(initial_value=tf.random.uniform(shape=(input_dim, output_dim)))
b = tf.Variable(initial_value=tf.zeros(shape = (output_dim, )))
```

```python
# forward pass
def model(inputs, W, b):
    return tf.matmul(inputs, W) + b
```

```python
def mean_sq_error(targets, predictions):
    # targets dim:
    # predictions dim:
    # print(f'targets dimension: {targets.shape}')
    # print(f'predictions: {predictions.shape}')
    per_sample_losses = tf.square(targets - predictions)

    # print(f'Per sample loss dimension: {per_sample_losses.shape}')
    return tf.reduce_mean(per_sample_losses)
```

```python
learning_rate = 0.1

def training_step(inputs, targets, W, b):
    with tf.GradientTape() as tape:
        predictions = model(inputs, W, b)
        loss = mean_sq_error(targets, predictions)
    grad_loss_wrt_W, grad_loss_wrt_b = tape.gradient(loss, [W,b])
    # print(f'W shape: {W.shape}')
    W.assign_sub(grad_loss_wrt_W * learning_rate)
    # print(f'b shape: {b.shape}')
    b.assign_sub(grad_loss_wrt_b * learning_rate)
    return loss # returned for metrics
```

```python
loss_ts = []

for step in range(40):
    loss = training_step(inputs, targets, W, b)
    loss_ts.append(loss)
    print(f'Loss at step {step}: {loss:.4f}')

print(loss_ts)
plt.plot(loss_ts)
plt.show()
```
